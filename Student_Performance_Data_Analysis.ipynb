# Student Performance Data Analysis

## Module 3 – Data Analysis & Visualization

This project demonstrates data analysis and visualization using NumPy, Pandas, and Matplotlib.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Maths": [55, 60, 65, 70, 75, 80, 85, 90],
    "Science": [60, 62, 68, 72, 78, 82, 88, 92],
    "English": [65, 68, 70, 75, 80, 84, 87, 94]
}

df = pd.DataFrame(data)

df
```

## 4. Explore Dataset


```python
print("First 5 rows:")
display(df.head())

print("Dataset information:")
df.info()

print("\nDataset statistics:")
display(df.describe())
```

## 5. Data Cleaning

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("\nDataset after cleaning:")
display(df)
```

## 6. Data Analysis

```python
print("Average scores:")

print("Maths:", df["Maths"].mean())
print("Science:", df["Science"].mean())
print("English:", df["English"].mean())

print("\nHighest Maths score:", df["Maths"].max())
print("Highest Science score:", df["Science"].max())
print("Highest English score:", df["English"].max())
```

## 7. Bar Chart - Average Scores

```python
subjects = ["Maths", "Science", "English"]

average_scores = [
    df["Maths"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

plt.bar(subjects, average_scores)

plt.title("Average Student Scores")
plt.xlabel("Subjects")
plt.ylabel("Average Score")

plt.show()
```

## 8. Line Chart - Study Hours vs Average Score

```python
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

plt.plot(df["Study_Hours"], df["Average"], marker="o")

plt.title("Study Hours vs Average Score")
plt.xlabel("Study Hours")
plt.ylabel("Average Score")

plt.grid(True)
plt.show()
```

## 9. Pie Chart - Subject Score Distribution

```python
subject_averages = [
    df["Maths"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

subjects = ["Maths", "Science", "English"]

plt.pie(
    subject_averages,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.title("Average Score Distribution by Subject")

plt.show()
```

## 10. Conclusion

The student performance dataset was successfully explored and cleaned using Pandas.
NumPy was used for numerical operations, and Matplotlib was used to create bar,
line, and pie charts.

The analysis helped us understand student performance and the relationship between
study hours and average scores.
