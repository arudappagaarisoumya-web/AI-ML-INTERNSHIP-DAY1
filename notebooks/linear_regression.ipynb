# Student Score Prediction Using Linear Regression

## 1. Introduction

This project uses Linear Regression to predict student scores based on study hours.

```python
from sklearn.linear_model import LinearRegression

print("Linear Regression Project")
```

## 2. Dataset

The dataset contains study hours and the corresponding student scores.

```python
# Study hours
X = [[1], [2], [3], [4], [5], [6], [7], [8]]

# Student scores
y = [35, 40, 50, 55, 60, 65, 75, 80]

print("Dataset loaded successfully!")
```

## 3. Model Training

The Linear Regression model is trained using the study hours and student scores.

```python
# Create the Linear Regression model
model = LinearRegression()

# Train the model using the dataset
model.fit(X, y)

print("Model trained successfully!")
```

## 4. Prediction

The trained model is used to predict the score for a given number of study hours.

```python
# Predict the score for 9 hours of study
prediction = model.predict([[9]])

print("Predicted score:", prediction[0])
```

## 5. Conclusion

The Linear Regression model was successfully trained using student study-hour data. The model can be used to predict student scores based on the number of hours studied.
